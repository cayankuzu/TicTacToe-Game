## **Tic-Tac-Toe: An Introduction**


---



---



Tic-Tac-Toe is a classic two-player game played on a 3x3 grid. Players take turns marking empty cells with their respective symbols, typically 'X' and 'O'.
The objective of the game is to align three of their symbols in a row, column, or diagonal.

### Rules of the Game:

---


1. The game is played on a 3x3 square grid.
2. The first player uses 'X', and the second player uses 'O'.
3. Players take turns placing their symbols in empty cells.
4. The player who aligns three symbols in a row, column, or diagonal wins the game.
5. If all cells are filled and no player has won, the game ends in a **draw**.

### Strategy:


---


Although the rules are simple, Tic-Tac-Toe requires players to make thoughtful moves. To increase their chances of winning, players must simultaneously work on completing their own rows while blocking their opponent’s attempts. This balance makes the game both fun and strategic.


### **2. Build the `Board` Class**

---


- **Set up the board**:
  - Use NumPy to create an empty board (e.g., `np.zeros` or `np.full`).
  - Allow the board size to be set when the game starts.

- **Add board functions**:
  - `place_symbol(row, col, symbol)`:
    - Put a symbol (X or O) in the selected cell.
    - Check if the move is valid (cell must be empty).
  - `is_full()`:
    - Check if the board is completely filled.
  - `check_winner()`:
    - Check for 3 in a row, column, or diagonal.
    - Use NumPy functions like `np.diagonal` or `np.all` for this.
  - `display_board()`:
    - Show the board in a clean way in the terminal.




````python
class Board:
    def __init__(self, size):
        self.board_size = size # size will be 3 which is size when we will creating game
        self.board = np.zeros((size, size), dtype=int) #creating zero matrix with givng size
    def place_symbol(self, row, col, symbol):
        if row < 0 or row >= self.board_size or col < 0 or col >= self.board_size: #checking if given symbol's (x or o) row and columns are in the matrix volume.
            print("Invalid move: Choose a cell within the board's boundaries.")
            return False
        if self.board[row, col] != 0:
            print("Invalid move: This cell is already occupied.") #ve create matrix with zeros. so this means if tehere is cell which is 0, is free.
            return False
        if symbol == 'X':
            self.board[row, col] = 1 #puttig 1 into cell which is player wants to put x. (don't wory, in display they see x)
        elif symbol == 'O':
            self.board[row, col] = -1 #puttig -1 into cell which is player wants to put x. (don't wory, in display they see o)
        return True
    def display_board(self):
        for row in self.board:
            for cell in row:
                if cell == 0:
                    print(" ", end="|")
                elif cell == 1:
                    print("X", end="|")
                else:
                    print("O", end="|")
            print("\n" + "-" * (2 * self.board_size - 1))
        print("\n")
    def is_full(self):
        return np.all(self.board != 0) #checking which cells is 0. this give us ture for 1 or -1 tre, false for 0
    def check_winner(self):
        for row in range(self.board_size): #self.board_size equal 3
            row_sum = np.sum(self.board[row, :]) #for exemple. takeing 1. row and sum their cells
            if row_sum == self.board_size: #if this sumation equals board_size which is 3
                return 1 #then return 1. so this means X wins. because X presents 1
            if row_sum == -self.board_size:  #if this sumation equals -(board_size) which is -3
                return -1 #then return -1. so this means O wins. because O presents  -1
        for col in range(self.board_size): #same operations for cols
            col_sum = np.sum(self.board[:, col]) #same operations for cols
            if col_sum == self.board_size:  #same operations for cols
                return 1 #same operations for cols
            if col_sum == -self.board_size: #same operations for cols
                return -1 #same operations for cols
        main_diag_sum = np.sum(np.diagonal(self.board)) #same operations for diagonals
        if main_diag_sum == self.board_size:  #same operations for diagonal
            return 1 #same operations for diagonal
        if main_diag_sum == -self.board_size:  #same operations for diagonal
            return -1 #same operations for diagonal
        anti_diag_sum = np.sum(np.diagonal(np.fliplr(self.board))) #same operations for reverse diagonal. let's look np.fliplr(self.board). here np.fliplr(self.board) means get revers of board columns. Now matrix reverse diagonals, turns into normal diagonals. So we can make normal diagonal summation.
        if anti_diag_sum == self.board_size:  #same operations for reverse diagonal
            return 1 #same operations for reverse diagonal
        if anti_diag_sum == -self.board_size:   #same operations for reverse diagonal
            return -1 #same operations for reverse diagonal
        return 0  #if there is no winer return 0
````

Summary: We create a 3x3 game table using a matrix. In this implementation, we can check who is the winner and whether the table is full.

- An important part is that we use numbers for X and O symbols. When checking the winner, we utilize matrix calculations. In column and row checks, we perform a basic sum operation on the matrix. The main idea behind this is: if any row or column is filled with X (whose numerical value is 1), the sum of this column or row must be 3, which is equal to our table size. Similarly, if it is filled with O (-1), their sum becomes -3, which is the negative value of our table size.

- Another important part is checking the diagonals, which include the main diagonal and the anti-diagonal. For the main diagonal, we can simply use numpy's diagonal summation. On the other hand, when checking the anti-diagonal, we can use the **fliplr()** function from numpy. The `np.fliplr()` function flips (inverts) a matrix horizontally, meaning the order of the matrix's columns is reversed. For example, check this:

### **3. Create the `Player` Class**

---


- **Store player information**:
  - `name`: The name of the player.
  - `symbol`: The player's symbol (X or O).

- **Keep it simple**:
  - This class only manages player info and switching turns.

````python
class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol
````


### **4. Create the `Game` Class**


---


- **Add functions for gameplay**:
  - `start_game()`:
    - Begin the game and control the rounds.
  - `play_turn()`:
    - Let players take turns to play.
  - `switch_player()`:
    - Alternate between Player X and Player O.
  - `check_winner()`:
    - Call the `check_winner()` function from the `Board` class.
  - `Check for a draw`:
    - If `is_full()` is true and no winner is found, the game ends in a draw.


- **Ask for player input**:
  - Get the row and column number where the player wants to place their symbol.
  - Make sure the input is valid (within bounds and not already taken).

- **Handle invalid moves**:
  - Show an error message if the move is invalid.
  - Let the player try again.

````python
class Game:
    def __init__(self, size): #this size == 3 (which we give while starting game)
        self.board = Board(size) # creating board object from Board class with (size)x(size)
        self.player1 = Player("Player 1", "X") #creating player1 from Player class
        self.player2 = Player("Player 2", "O") #creating player2 from Player class
        self.current_player = self.player1 # setting player who will first play when game start
    def start_game(self):
        while True:
            self.board.display_board()  #show the board from Board class (because self.board = Board(size))
            print(f"\n{self.current_player.name}'s ({self.current_player.symbol}) Turn") #printing which player turns
            try:
                row = int(input("Row (1-3): ")) #getting row's number which is will fill player's symbol
                col = int(input("Column (1-3): ")) #getting row's number which is will fill player's symbol
            except ValueError:
                print("Invalid input, please enter a number.")
                continue
            if self.board.place_symbol(row-1, col-1, self.current_player.symbol): #placing player's symbol to wanted cell.
                winner = self.board.check_winner() # checking if there is a winner quick after then player put a symbol in cell.
                if winner != 0: #this means every time after the check,
                    self.board.display_board() #show board
                    if winner == 1: #this means if winner is x. because from self.board.check_winner give us 1 when x wins
                        print(f"Winner: {self.player1.name} (X)")
                    else:
                        print(f"Winner: {self.player2.name} (O)") #this means winner is o when self.board.check_winner give us -1. because there is no another option.
                    break
                if self.board.is_full(): #this means there is no another free cell.
                    self.board.display_board() #show board
                    print("It's a draw!") #its draw becaus no winner
                    break
                self.switch_player() #its really whitching player while game is on
    def switch_player(self):
        if self.current_player == self.player1: #setting current player to player1
            self.current_player = self.player2 #ad quick after, setting current player to player2
        else:
            self.current_player = self.player1 #if current player is player 2, then setting current player to layer 1. in summary we create cycle here.
````

Form here we use while loop because we want to our game on while some player win or draw. The logic of this while loop is like Pygame.


### **5. Final Steps**

---


- Sart the game (:

````python
game = Game(3) #creating game from Game class with set size of board
game.start_game() #starting game from Game class
````
